````python
from huggingface_hub import InferenceClient
import os
from dotenv import load_dotenv

load_dotenv()

client = InferenceClient(
    api_key=os.getenv("HF_TOKEN")
)

try:
    completion = client.chat.completions.create(
        model="XiaomiMiMo/MiMo-V2.5:deepinfra",
        messages=[
            {"role": "user", "content": "What is the capital of France?"}
        ],
    )

    print(completion.choices[0].message.content)

except Exception as e:
    print(type(e).__name__)
    print(e)
````

# LLM Generation Parameters

## Overview

Large Language Models (LLMs) generate text one token at a time. At each
step, the model predicts probabilities for the next token, and
generation parameters influence **how the next token is selected**.

------------------------------------------------------------------------

# 1. `max_tokens`

**Purpose:** Limits the maximum number of tokens the model can generate.

``` python
max_tokens=200
```

-   Short replies: **20--100**
-   Chatbots: **200--500**
-   Stories: **500--2000**
-   Code generation: **500+**

------------------------------------------------------------------------

# 2. `temperature`

**Purpose:** Controls randomness and creativity.

Lower values make the output more deterministic.

Higher values make the output more diverse and creative.

    Temperature Behavior
  ------------- ----------------------
            0.0 Deterministic
            0.2 Mostly deterministic
            0.7 Balanced
            1.0 Creative
           1.5+ Highly unpredictable

### Example

``` python
temperature=0
```

> Best for coding, math, factual Q&A.

``` python
temperature=1.2
```

> Better for storytelling and brainstorming.

------------------------------------------------------------------------

# 3. `top_p`

**Purpose:** Uses *nucleus sampling*.

Instead of considering every possible next token, the model only
considers the smallest group of tokens whose cumulative probability
reaches `top_p`.

``` python
top_p=0.9
```

-   Lower values → Conservative outputs
-   Higher values → More diverse outputs

Typical values:

-   `0.8`
-   `0.9`
-   `0.95`
-   `1.0`

> In practice, adjust either **temperature** or **top_p** significantly,
> rather than both at once.

------------------------------------------------------------------------

# 4. `frequency_penalty`

**Purpose:** Discourages repeating the same words or phrases.

``` python
frequency_penalty=0.8
```

Useful for:

-   Essays
-   Long-form writing
-   Stories
-   Summaries

Range:

    -2.0 to 2.0

Higher values reduce repetition.

------------------------------------------------------------------------

# 5. `presence_penalty`

**Purpose:** Encourages introducing new topics and vocabulary.

``` python
presence_penalty=0.6
```

Useful when you want the model to explore new ideas instead of
revisiting the same concepts.

Range:

    -2.0 to 2.0

------------------------------------------------------------------------

# 6. `stop`

**Purpose:** Stops generation when a specified string is encountered.

``` python
stop=["END"]
```

Example:

Model generates

    Hello
    END
    More text...

Returned output

    Hello

Useful for structured outputs, templates, and conversations.

------------------------------------------------------------------------

# 7. `stream`

**Purpose:** Streams tokens as they are generated.

``` python
stream=True
```

Instead of waiting for the full response, text arrives token by token.

Useful for:

-   Chat applications
-   Interactive UIs
-   Real-time assistants

------------------------------------------------------------------------

# 8. `seed`

**Purpose:** Makes generation more reproducible.

``` python
seed=42
```

Useful for:

-   Benchmarking
-   Debugging
-   Prompt engineering
-   Experiments

------------------------------------------------------------------------

# Recommended Settings

| Task             | Temperature |   Top_p | Max Tokens |<br>
| ---------------- | ----------: | ------: | ---------: |<br>
| Coding           |       0–0.2 |     1.0 |   500–2000 |<br>
| Math             |           0 |     1.0 |   200–1000 |<br>
| Q&A              |     0.2–0.5 | 0.9–1.0 |    200–500 |<br>
| General chatbot  |         0.7 |     0.9 |    200–800 |<br>
| Creative writing |     0.9–1.3 |    0.95 |   500–2000 |<br>
| Storytelling     |         1.0 |    0.95 |  1000–3000 |<br>


------------------------------------------------------------------------

# Example

``` python
completion = client.chat.completions.create(
    model="XiaomiMiMo/MiMo-V2.5:deepinfra",
    messages=[
        {
            "role": "system",
            "content": "You are a helpful AI tutor."
        },
        {
            "role": "user",
            "content": "Explain transformers in simple terms."
        }
    ],
    max_tokens=300,
    temperature=0.7,
    top_p=0.9,
    frequency_penalty=0.5,
    presence_penalty=0.2,
    seed=42,
)
```

------------------------------------------------------------------------

# Key Takeaways

-   **max_tokens** → Output length
-   **temperature** → Randomness
-   **top_p** → Candidate token filtering
-   **frequency_penalty** → Reduce repetition
-   **presence_penalty** → Encourage new topics
-   **stop** → End generation on specific text
-   **stream** → Token-by-token output
-   **seed** → Reproducible generations
