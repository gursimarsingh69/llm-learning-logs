```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df= pd.read_csv(r"..\assets\SpotifySongs.csv")
df.shape
df.info()
df.columns
df.describe().T   
```

```python
features = [
    "Popularity",
    "Danceability",
    "Energy",
    "Key",
    "Loudness",
    "Mode",
    "Speechiness",
    "Acousticness",
    "Instrumentalness",
    "Liveness",
    "Valence",
    "Tempo",
    "Duration_ms"
]

X = df[features]

X.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler() #it calculates z score z=(x−μ​)/σ
 
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix=cosine_similarity(X_scaled)

similarity_matrix.shape
```

```python
song_to_index = pd.Series(df.index, index=df["SongName"]).drop_duplicates()
```

```python
song_to_index["Higher Power"]
```

```python
song_index = song_to_index["Saiyaan Ji"]

similarity_scores = similarity_matrix[song_index]

similarity_scores[:10]
```

```python
list(enumerate(similarity_scores))
```

```python
sorted_scores = sorted(
    list(enumerate(similarity_scores)),
    key=lambda x: x[1],
    reverse=True
)
```

```python
top_recommendations = sorted_scores[1:6]
sorted_scores
```

```python
for index, score in top_recommendations:
    print(df.iloc[index]["SongName"], score)
```
